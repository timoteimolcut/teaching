## Linked Lists

```python
# building block of a list
# (), (()), ((())), (((()))) -> inheritance, matroska dolls, recursion
# [(), ()], [(), (), (), (), ()] -> composition, aggregation, has-a relationship
class Node:
    # dunder methods: __init__, __str__, __repr__, __eq__, etc.
    def __init__(self, value): # constructor
        self.value = value # data
        self.next = None # pointer to the next node
    def __repr__(self): # representation: how do I want to see the object when I print it?
        return f'Node({self.value})'
# print(type(None)) # zero for objects, empty for lists, dicts, etc.


n1 = Node(4) # first node
n2 = Node(8) # second node
n3 = Node(16) # third node
n1.next = n2 # n1 points to n2, n2 is next of n1
n2.next = n3 # n2 points to n3, n3 is next of n2
# list: head -> [n1:4] -> [n2:8] -> [n3:16] -> None

print('linked list:', n1.value, end=' ')
print(n1.next.value, end=' ')
print(n1.next.next.value, end=' ')

# mirror example with built-in list
l = list()
l.append(4)
l.append(8)
l.append(16)
l
```

```python
#1
class Double_Node: # creating the node from scratch
    # dunder methods: __init__, __str__, __repr__, __eq__, etc.
    def __init__(self, value): # constructor
        self.value = value # data
        self.next = None # pointer to the next node
        self.prev = None # pointer to the previous node
    def __repr__(self): # representation: how do I want to see the object when I print it?
        return f'D_Node({self.value})'
#2
class DoubleLinkedListNode(Node): # we use inheritance: we use out predefined Node (single linked lists) and we augment to previous link as well
    def __init__(self, value):
        super().__init__(value)
        self.prev = None
    # def __repr__(self):
    #     return f'DoubleLinkedListNode({self.value})'
# inference is when we run a model
# inheritance from parent class to child class
# instance is when we "spawn" an object from a class 

dl_n1 = DoubleLinkedListNode(4)
print(dl_n1)
d_n1 = Double_Node(4)
print(d_n1)
```

```python
class DoubleLinkedList:
    def __init__(self):
        self.head = None
    def __repr__(self):
        return f'DoubleLinkedList({self.head})'
    def insert(self, value, idx):
        """
        Insert a new node containing value, at the fixed position of idx.
        return None
        """
        new_node = DoubleLinkedListNode(value)
        # we as programmers need to cover all the possible cases
        if idx < 0: # but we can develop in the future ex. list[-1] = ...
            print("Only positive indexes allowed.")
            return
        # idx = 0
        # idx = len -1 - last element
        # idx is in between
        # case 1
        if idx == 0:
            # case 1.1 this case: empty list
            if self.head is None:
                self.head = new_node    
                return
            # case 1.2 this case: the list is not empty 
            # attaching new node to the whole list
            new_node.next = self.head
            self.head.prev = new_node
            # update the head
            self.head = new_node
            return
        # case 2 idx > 0
        count = 0
        curr = self.head
        # we want to reach the node is at the conceptual position of idx
        # but we don't want to get out of the list
        while curr is not None and count < idx: # both have to be true, always
            curr = curr.next
            count += 1
        # count == idx
        if curr is None:
            print("idx is out of range")
            return
        # count == idx and curr is before the end of the list
        # here curr is the last element so we insert at the end of the list
        # curr is inside the list: after head and before the last
        # curr.next
        # curr.next.prev
        new_node.next = curr
        new_node.prev = curr.prev
        curr.prev.next = new_node
        curr.prev = new_node
        

    def append(self, value):
        new_node = DoubleLinkedListNode(value)
        # the list is empty
        if self.head is None:
            self.head = new_node
            return
        # not empty
        curr = self.head # current node
        while(curr.next is not None): # we want to reach the last node, but not exit the list
            curr = curr.next
        curr.next = new_node
    def traverse(self):# only left to rigth 
        print("DoubleLinkedList: ", end=" ")
        curr = self.head
        while(curr is not None):
            if curr.next is None:
                print(curr)
                return
            print(curr, end=", ")
            curr = curr.next

```

```python
dll = DoubleLinkedList()
dll.insert(123, 0)
dll.insert(234, 0)
dll.insert(345, 0)
dll.insert(456, 0)
dll.insert(1, 2)

dll.traverse()
```
