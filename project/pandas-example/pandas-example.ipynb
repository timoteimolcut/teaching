```python
import pandas as pd

df = pd.read_csv("employees.csv")
df.head()
```

```python
df.info()
df.describe()
df.columns
```

```python
df["Name"]
df[["Name", "Salary"]]
```

```python
df[df["Department"] == "IT"]
df[df["Salary"] > 6000]
```

```python
df["Total"] = df["Salary"] + df["Bonus"]
df.head()
```

```python
df.groupby("Department")["Total"].mean()
```

```python
df.groupby("Department").agg({
    "Salary": "mean",
    "Bonus": "sum",
    "Total": "max"
})
```

```python
df.sort_values(by="Salary", ascending=False)
```

```python
df.nlargest(5, "Salary")
```

```python
def tax(salary):
    return salary * 0.1

df["Tax"] = df["Salary"].apply(tax)
df.head()
```

```python
df.to_csv("output.csv", index=False)
```
